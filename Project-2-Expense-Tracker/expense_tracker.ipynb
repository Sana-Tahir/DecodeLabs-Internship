```python
total = 0

print("Enter 'done' when you are finished.")

while True:
    expense = input("Enter expense: ")

    if expense.lower() == "done":
        print("Bye!")
        break

    try:
       expense = float(expense)
       total += expense

    except ValueError:
        print("Invalid output")

print("Total Spent:", total)
```
