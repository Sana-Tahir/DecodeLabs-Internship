```python
task = []

while True:
    print("1. Add Task")
    print("2. View Task")
    print("3. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        user = input("Enter your task: ")
        task.append(user)
        print("Task Added!")

    elif choice == "2":
        if len(task) == 0:
            print("No tasks available.")
        else:
            print("\nYour Tasks:")
            for index, tasks in enumerate(task, start=1):
                print(f"{index}. {tasks}")

    elif choice == "3":
        print("Bye")
        break

    else:
        print("Invalid choice")
```
